```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Deux graphiques côte à côte
fig, axes = plt.subplots(1, 2, figsize=(18, 6))

# Graphique de gauche avant 1951
ax1 = axes[0]
ax2 = axes[1]

x: int = np.linspace(0, 10, 2)
b: int = 0

# Premier graphique avec a négatif
a: int = -1
y: int = a * x + b

# Second graphique avec a positif
c: int = 1
z: int = c * x + b


g = np.cos(x)

plt.plot(x, g)

ax1.plot(x, y, linewidth=2, color='red', linestyle='--')
ax2.plot(x, z, linewidth=2, color='green', linestyle='--')

# ligne à 0 pour référence
ax1.axhline(0, color='black', linewidth=2, alpha=1)
ax2.axhline(0, color='black', linewidth=2, alpha=1)

# Titres et labels
ax1.set_title('Représentation graphique de ax + b avec a négatif', fontsize=10, fontweight='bold')
ax1.set_xlabel('Valeur de x', fontsize=12)
ax1.set_ylabel('Valeur de y', fontsize=12)
ax1.legend(fontsize=10)
ax1.grid(True, alpha=0.3)

ax2.set_title('Représentation graphique de ax + b avec a positif', fontsize=10, fontweight='bold')
ax2.set_xlabel('Valeur de x', fontsize=12)
ax2.set_ylabel('Valeur de y', fontsize=12)
ax2.legend(fontsize=10)
ax2.grid(True, alpha=0.3)

plt.show() # affiche la figure à l'écran
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 10, 2)
b = 0

a = 1
y = x * a + b

c = -1
z = x * c + b

fig, ax = plt.subplots()
ax.plot(x, y, linewidth=2, color='green', linestyle='--')
ax.plot(x, z, linewidth=2, color='red', linestyle='-')

ax.axhline(0, color='black', linewidth=2, alpha=1)

plt.show() # affiche la figure à l'écran
```
